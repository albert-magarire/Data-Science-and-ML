##Part 1

Follow the instructions while reading the project.  Our goal is randomly generate matrices $A$ and $b$ that have a single solution $x$ for the equation $$Ax=b.$$

1. Generate a new random matrix with 5 rows and 10 columns with integer values between -10 and 100.  Call this matrix $A$.

```python
#Run this cell with appropriate changes

rows_A, cols_A = 5, 10 #An example: a 5x10 matrix

low_A, high_A = -10, 100 #These numbers are going to be the bounds for our random integers.  The low is -10 and the high is 100.

#Generate a random integer matrix

A = np.random.randint(low_A, high_A, size=(rows_A, cols_A))

print(A)
```

2. For this matrix $A$, generate a random column vector, $b$ with integer values between -20 and 15.  The matrix $A$ will help you determine the number of rows for $b$. There should only be 1 column.  

```python
#Run this cell

rows_b, cols_b = 5,1
low_b,high_b = -20,15

b= np.random.randint(low_b, high_b,size=(rows_b,cols_b))

print(b)
```

3. Solve the system using the x = np.linalg.solve(A,b) command.

```python
#Run this cell

x = np.linalg.solve(A,b)
```

We get an error! That's absurd. Okay, not really.  We get this error because numpy.linalg.solve uses a method of solving that requires a square matrix.  We have a way around this, multiply both sides of the equation \begin{align*}
Ax &= b
\end{align*} by $A^T$ on the **left**.  Note that if $A$ is a $m\times n$ matrix, then $A^T$ is a $n\times m$ matrix.  Hence, $A^TA$ will be a $n\times n$ matrix. This gives us the equation \begin{align*} (A^TA)x = A^Tb. \end{align*} Now we can use numpy.linalg.solve.

I believe that numpy.linalg.solve requires the use of a square matrix because it wants to multiply both sides by the *inverse* of $A$.  To get $$x=A^{-1}b.$$

When we use numpy.linalg.solve to solve our **new** matrix equation, we are actually finding $$x = (A^TA)^{-1}(A^Tb).$$  In a perfect world, one where $A$ is a square and invertible matrix, $(A^TA)^{-1} = A^{-1}(A^T)^{-1}$.  So our solution would be $$x = A^{-1}b,$$ but our $A$ is not square.  Thus, it is not invertible.  So we actually have that $(A^TA)^{-1} \ne A^{-1}(A^T)^{-1}$ since $A^{-1}$ does not exist.

4. Find $A^T$, label it A_T, and use numpy.linalg.solve to find $x$ that satisfies the equation $$A^TAx=A^Tb.$$

```python
#First we define the transpose

A_T = np.transpose(A)

#Next we solve the (A_T@A)x = (A_T)b

x = np.linalg.solve(A_T@A,A_T@b)

print(x)
```

5. If you get that there is no solution since the matrix is singular, this means that $$A^TA$$ is not invertible.  Regenerate $A$ and your code until you get one solution. Go back to the first item and rerun the cells.

6. We can check to make sure that the $x$ we found also satisfies the equation $$Ax=b$$ for our randomly generated matrices by using the numpy.allclose function and defining a tolerance level.

```python
#Run this cell

#Define a tolerance level

tolerance = 1e-5 #tolerance is 0.00001

are_equal = np.allclose(A@x,b, atol = tolerance) #atol means absolute tolerance

print(are_equal)
```
